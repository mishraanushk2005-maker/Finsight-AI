```python
df["date"] = pd.to_datetime(df["date"])

numeric_cols = [
    "revenue", "expense", "marketing_expense", "employee_cost",
    "operations_cost", "budget_revenue", "budget_expense"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.drop_duplicates()
df = df.sort_values(["date", "company", "department"]).reset_index(drop=True)
```

```python
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.drop_duplicates()
df = df.sort_values(["date", "company", "department"]).reset_index(drop=True)
```

```python
# Recalculate important business metrics
df["profit"] = df["revenue"] - df["expense"]
df["profit_margin"] = (df["profit"] / df["revenue"]) * 100
df["revenue_variance"] = df["revenue"] - df["budget_revenue"]
df["revenue_variance_pct"] = (df["revenue_variance"] / df["budget_revenue"]) * 100
df["expense_variance"] = df["expense"] - df["budget_expense"]
df["expense_variance_pct"] = (df["expense_variance"] / df["budget_expense"]) * 100

print("After cleaning:", df.shape)
print("Missing values:", df.isnull().sum().sum())
df.head()
```

```python
from statsmodels.tsa.stattools import adfuller
def adf_test(series, name):
    result = adfuller(series.dropna())
    print(f"ADF test: {name}")
    print("ADF statistic:", result[0])
    print("p-value:", result[1])
    print("Critical values:")
    for key, value in result[4].items():
        print(f"  {key}: {value}")
    if result[1] < 0.05:
        print("Conclusion: likely stationary (reject H0).")
    else:
        print("Conclusion: likely non-stationary (fail to reject H0).")

adf_test(monthly["revenue"], "Original Revenue")
```

```python
monthly["revenue_diff"] = monthly["revenue"].diff()

adf_test(monthly["revenue_diff"], "First-Differenced Revenue")
```

```python
plt.figure(figsize=(15,5))
plt.plot(monthly["date"], monthly["revenue_diff"])
plt.title("First-Differenced Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue Difference")
plt.grid(alpha=0.25)
plt.show()
```

```python
series = monthly[["date", "revenue"]].copy()

TEST_SIZE = 12

train = series.iloc[:-TEST_SIZE].copy()
test = series.iloc[-TEST_SIZE:].copy()

print("Train:", train["date"].min().date(), "to", train["date"].max().date())
print("Test :", test["date"].min().date(), "to", test["date"].max().date())
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
# Predict every future month using the last observed training value
import numpy as np
naive_pred = np.repeat(train["revenue"].iloc[-1], len(test))

naive_mae = mean_absolute_error(test["revenue"], naive_pred)
naive_rmse = np.sqrt(mean_squared_error(test["revenue"], naive_pred))

print("Naive MAE :", naive_mae)
print("Naive RMSE:", naive_rmse)
```

```python
# Since revenue has yearly seasonality, use the value from 12 months earlier.
seasonal_naive_pred = train["revenue"].iloc[-12:].values

seasonal_mae = mean_absolute_error(test["revenue"], seasonal_naive_pred)
seasonal_rmse = np.sqrt(mean_squared_error(test["revenue"], seasonal_naive_pred))

print("Seasonal Naive MAE :", seasonal_mae)
print("Seasonal Naive RMSE:", seasonal_rmse)
```

```python
from statsmodels.tsa.arima.model import ARIMA
```

```python
# A simple ARIMA baseline.
# The LSTM should beat this only if the test results actually show that.

arima_model = ARIMA(train["revenue"], order=(1,1,1))
arima_fit = arima_model.fit()

arima_pred = arima_fit.forecast(steps=TEST_SIZE)

arima_mae = mean_absolute_error(test["revenue"], arima_pred)
arima_rmse = np.sqrt(mean_squared_error(test["revenue"], arima_pred))

print("ARIMA MAE :", arima_mae)
print("ARIMA RMSE:", arima_rmse)
```

```python
comparison = pd.DataFrame({
    "date": test["date"].values,
    "actual": test["revenue"].values,
    "naive": naive_pred,
    "seasonal_naive": seasonal_naive_pred,
    "arima": np.asarray(arima_pred)
})

display(comparison)
```

```python
plt.figure(figsize=(15,6))
plt.plot(train["date"].iloc[-36:], train["revenue"].iloc[-36:], label="Training History")
plt.plot(test["date"], test["revenue"], marker="o", label="Actual")
plt.plot(test["date"], naive_pred, label="Naive")
plt.plot(test["date"], seasonal_naive_pred, label="Seasonal Naive")
plt.plot(test["date"], arima_pred, label="ARIMA")
plt.title("Baseline Forecast Comparison")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
LOOKBACK = 12

train_values = train["revenue"].values.reshape(-1, 1)
test_values = test["revenue"].values.reshape(-1, 1)

scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_values)

# Add the last LOOKBACK training values before the test period
combined = np.concatenate([train_values[-LOOKBACK:], test_values])
combined_scaled = scaler.transform(combined)

X_test, y_test = [], []

for i in range(LOOKBACK, len(combined_scaled)):
    X_test.append(combined_scaled[i-LOOKBACK:i, 0])
    y_test.append(combined_scaled[i, 0])

X_test = np.array(X_test)
y_test = np.array(y_test)

X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
# Create supervised sequences from training data
X_train, y_train = [], []

for i in range(LOOKBACK, len(train_scaled)):
    X_train.append(train_scaled[i-LOOKBACK:i, 0])
    y_train.append(train_scaled[i, 0])

X_train = np.array(X_train)
y_train = np.array(y_train)

X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
tf.random.set_seed(42)

model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(LOOKBACK, 1)),
    Dropout(0.2),

    LSTM(32),
    Dropout(0.2),

    Dense(16, activation="relu"),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

model.summary()
```

```python
early_stop = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.15,
    shuffle=False,
    callbacks=[early_stop],
    verbose=1
)
```

```python
plt.figure(figsize=(10,5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.title("LSTM Training History")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
pred_scaled = model.predict(X_test, verbose=0)

lstm_pred = scaler.inverse_transform(pred_scaled).flatten()
actual = test["revenue"].values

lstm_mae = mean_absolute_error(actual, lstm_pred)
lstm_rmse = np.sqrt(mean_squared_error(actual, lstm_pred))

print("LSTM MAE :", lstm_mae)
print("LSTM RMSE:", lstm_rmse)
```

```python
results = pd.DataFrame({
    "date": test["date"].values,
    "actual": actual,
    "lstm_prediction": lstm_pred
})

display(results)
```

```python
plt.figure(figsize=(15,6))
plt.plot(test["date"], actual, marker="o", label="Actual")
plt.plot(test["date"], lstm_pred, marker="o", label="LSTM")
plt.title("LSTM — 12-Month Test Forecast")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
metrics = pd.DataFrame({
    "Model": ["Naive", "Seasonal Naive", "ARIMA", "LSTM"],
    "MAE": [naive_mae, seasonal_mae, arima_mae, lstm_mae],
    "RMSE": [naive_rmse, seasonal_rmse, arima_rmse, lstm_rmse]
}).sort_values("RMSE")

display(metrics)
```

```python
all_values = series["revenue"].values.reshape(-1, 1)

final_scaler = MinMaxScaler()
all_scaled = final_scaler.fit_transform(all_values)

X_all, y_all = [], []

for i in range(LOOKBACK, len(all_scaled)):
    X_all.append(all_scaled[i-LOOKBACK:i, 0])
    y_all.append(all_scaled[i, 0])

X_all = np.array(X_all).reshape(-1, LOOKBACK, 1)
y_all = np.array(y_all)

tf.random.set_seed(42)

final_model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(LOOKBACK, 1)),
    Dropout(0.2),
    LSTM(32),
    Dropout(0.2),
    Dense(16, activation="relu"),
    Dense(1)
])

final_model.compile(optimizer="adam", loss="mse")

final_model.fit(
    X_all, y_all,
    epochs=60,
    batch_size=16,
    validation_split=0.1,
    shuffle=False,
    callbacks=[EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True)],
    verbose=1
)
```

```python
future_steps = 12

last_window = all_scaled[-LOOKBACK:].flatten().tolist()
future_scaled = []

for _ in range(future_steps):
    x_input = np.array(last_window[-LOOKBACK:]).reshape(1, LOOKBACK, 1)
    next_scaled = final_model.predict(x_input, verbose=0)[0, 0]

    future_scaled.append(next_scaled)
    last_window.append(next_scaled)

future_revenue = final_scaler.inverse_transform(
    np.array(future_scaled).reshape(-1, 1)
).flatten()

future_dates = pd.date_range(
    start=series["date"].max() + pd.offsets.MonthBegin(1),
    periods=future_steps,
    freq="MS"
)

forecast_12m = pd.DataFrame({
    "date": future_dates,
    "forecast_revenue": future_revenue
})

display(forecast_12m)
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    series["date"].iloc[-36:],
    series["revenue"].iloc[-36:],
    label="Historical Revenue"
)

plt.plot(
    forecast_12m["date"],
    forecast_12m["forecast_revenue"],
    marker="o",
    linewidth=2,
    label="LSTM 12-Month Forecast"
)

plt.title("Financial Revenue — 12-Month LSTM Forecast")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```
